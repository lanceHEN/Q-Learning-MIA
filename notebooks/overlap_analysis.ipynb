```python
continuous_csvs = ["../data/tables/overlap_acrobot_cql_epsilon_greedy.csv"]
continuous_titles = ["Acrobot (Epsilon-greedy)"]
```

```python
def plot_overlaps_continuous(csv, title):
    df = pd.read_csv(csv)
    df = df.groupby("N. Trajectories").agg({"Nearest Distance Sum": sum, "Nonmember Pairs": sum})
    df["Mean Min. Pairwise Dist"] = df["Nearest Distance Sum"] / df["Nonmember Pairs"]
    plt.plot(df.index, df["Mean Min. Pairwise Dist"])
    plt.xlabel("n")
    plt.ylabel("Mean Minimum Pairwise Distance")
    plt.title(title)

for csv, title in zip(continuous_csvs, continuous_titles):
    plot_overlaps_continuous(csv, title)
```
